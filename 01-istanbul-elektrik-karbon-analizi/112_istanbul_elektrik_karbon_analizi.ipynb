## 7. Veri Setinin Yapısal İncelemesi

## 8. Veri Kalitesi ve Temizleme

Veri analizinin güvenilirliğini sağlamak amacıyla veri setindeki eksik değerler, tekrarlanan kayıtlar ve geçersiz tüketim miktarları kontrol edilmiştir. Ayrıca yıllara göre dönemsel bütünlük incelenmiş ve farklı yıllarda kullanılan tüketici grubu adları analiz için standartlaştırılmıştır.

Olağan dışı tüketim değerlerini belirlemek amacıyla sektör bazında çeyrekler arası açıklık (IQR) yöntemi uygulanmıştır. Belirlenen aykırı değerler, önceki ve sonraki aylardaki tüketim miktarlarıyla karşılaştırılarak grafikler üzerinde incelenmiştir.

Aykırı değerler gerçek tüketim değişimlerini yansıtabileceğinden, kaynak verilerden herhangi bir gözlem silinmemiş veya değiştirilmemiştir.

```python
print("Eksik değer sayıları:")
print(df.isnull().sum())

print("\nTekrarlanan kayıt sayısı:")
print(df.duplicated(subset=["yil", "ay", "tuketici_turu"]).sum())

print("\nNegatif tüketim sayısı:")
print((df["elektrik_tuketimi_mwh"] < 0).sum())

print("\nSıfır tüketim sayısı:")
print((df["elektrik_tuketimi_mwh"] == 0).sum())
```

```python
# Yıllara göre ay sayısı
print("Yıllara göre ay sayısı:")
print(df.groupby("yil")["ay"].nunique())

# Her yıl ve ay için tüketici grubu sayısı
print("\nAy başına tüketici grubu sayıları:")
print(
    df.groupby(["yil", "ay"])["tuketici_turu"]
    .nunique()
    .value_counts()
)
```

```python
print("Standart tüketici grupları:")
print(df["tuketici_turu"].unique())

print("\nEPDK'daki orijinal tüketici türleri:")
print(df["epdk_orijinal_tur"].unique())

print("\nYıllara göre orijinal tüketici türleri:")
print(
    df.groupby("yil")["epdk_orijinal_tur"]
    .unique()
)
```

```python
# Orijinal veri setini koruyarak çalışma kopyası oluşturma
df_temiz = df.copy()

# Tüketici gruplarını standartlaştırma
grup_esleme = {
    "Tarım": "Tarım",
    "Tarımsal Faaliyetler": "Tarım",
    "Hizmetler/Ticarethane": "Hizmetler/Ticarethane",
    "Kamu ve Özel Hizmetler Sektörü ile Diğer": "Hizmetler/Ticarethane"
}

df_temiz["tuketici_turu"] = (
    df_temiz["tuketici_turu"]
    .replace(grup_esleme)
)

print("Standartlaştırma sonrası tüketici grupları:")
print(df_temiz["tuketici_turu"].unique())

print("\nGrup sayısı:", df_temiz["tuketici_turu"].nunique())
```

```python
# Sektör bazında IQR yöntemiyle aykırı değer kontrolü
def aykiri_deger_bul(grup):
    Q1 = grup["elektrik_tuketimi_mwh"].quantile(0.25)
    Q3 = grup["elektrik_tuketimi_mwh"].quantile(0.75)
    IQR = Q3 - Q1

    alt_sinir = Q1 - 1.5 * IQR
    ust_sinir = Q3 + 1.5 * IQR

    return grup[
        (grup["elektrik_tuketimi_mwh"] < alt_sinir) |
        (grup["elektrik_tuketimi_mwh"] > ust_sinir)
    ]

aykiri_degerler = pd.concat([
    aykiri_deger_bul(grup)
    for _, grup in df_temiz.groupby("tuketici_turu")
])

print("Aykırı değer sayısı:", len(aykiri_degerler))

display(
    aykiri_degerler[
        ["yil", "ay", "tuketici_turu", "elektrik_tuketimi_mwh"]
    ].sort_values(["tuketici_turu", "yil", "ay"])
)
```

```python
# Aykırı değerlerin önceki ve sonraki aylarla karşılaştırılması
df_temiz["tarih"] = pd.to_datetime(
    dict(year=df_temiz["yil"], month=df_temiz["ay"], day=1)
)

df_temiz = df_temiz.sort_values(
    ["tuketici_turu", "tarih"]
)

df_temiz["onceki_ay"] = (
    df_temiz.groupby("tuketici_turu")["elektrik_tuketimi_mwh"]
    .shift(1)
)

df_temiz["sonraki_ay"] = (
    df_temiz.groupby("tuketici_turu")["elektrik_tuketimi_mwh"]
    .shift(-1)
)

display(
    df_temiz.loc[
        aykiri_degerler.index,
        ["tarih", "tuketici_turu", "onceki_ay",
         "elektrik_tuketimi_mwh", "sonraki_ay"]
    ].sort_values(["tuketici_turu", "tarih"])
)
```

```python
# Sektör bazında kutu grafikleri
sektorler = df_temiz["tuketici_turu"].unique()

fig, axes = plt.subplots(1, 5, figsize=(17, 5))

for ax, sektor in zip(axes, sektorler):
    veriler = df_temiz.loc[
        df_temiz["tuketici_turu"] == sektor,
        "elektrik_tuketimi_mwh"
    ]

    ax.boxplot(veriler)
    ax.set_title(sektor, fontsize=10)
    ax.set_xticks([])
    ax.set_ylabel("MWh")
    ax.grid(axis="y", alpha=0.3)

plt.suptitle("Sektörlere Göre Elektrik Tüketimi Dağılımı")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(5, 1, figsize=(13, 16))

for ax, (sektor, grup) in zip(
    axes,
    df_temiz.groupby("tuketici_turu")
):
    ax.plot(
        grup["tarih"],
        grup["elektrik_tuketimi_mwh"],
        marker="o",
        markersize=3
    )

    aykiri = aykiri_degerler[
        aykiri_degerler["tuketici_turu"] == sektor
    ]

    ax.scatter(
        df_temiz.loc[aykiri.index, "tarih"],
        aykiri["elektrik_tuketimi_mwh"],
        color="red",
        s=45,
        label="IQR aykırı değerleri",
        zorder=3
    )

    ax.set_title(sektor)
    ax.set_ylabel("MWh")
    ax.grid(alpha=0.3)
    ax.legend()

plt.tight_layout()
plt.show()
```

## 9. Keşifsel Veri Analizi (EDA)

Veri temizleme ve kalite kontrollerinin ardından İstanbul'un sektörel elektrik tüketimindeki eğilimleri belirlemek amacıyla keşifsel veri analizi gerçekleştirilecektir.

Bu kapsamda elektrik tüketiminin temel istatistikleri hesaplanacak; tüketici gruplarının tüketim düzeyleri, yıllık değişimler ve aylık tüketim örüntüleri karşılaştırılacaktır. Ayrıca sektörlerin toplam tüketimdeki payları ve mevsimsel farklılıklar grafiklerle incelenecektir.

2026 yılı yalnızca Ocak–Ağustos dönemini kapsadığından, yıllık karşılaştırmalarda tamamlanmış yıllar esas alınacak; 2026 verileri önceki yılların aynı dönemleriyle değerlendirilecektir.

```python
# Elektrik tüketiminin temel istatistikleri
display(
    df_temiz["elektrik_tuketimi_mwh"].describe()
)

# Sektör bazında temel istatistikler
display(
    df_temiz.groupby("tuketici_turu")["elektrik_tuketimi_mwh"]
    .agg(["count", "mean", "median", "min", "max", "std"])
    .round(2)
)
```

```python
# Tamamlanmış yılların toplam elektrik tüketimi
yillik_tuketim = (
    df_temiz[df_temiz["yil"] <= 2025]
    .groupby("yil")["elektrik_tuketimi_mwh"]
    .sum()
)

display(yillik_tuketim.round(2))
```

```python
# Yıllık elektrik tüketimi ve değişim oranı
yillik_analiz = yillik_tuketim.to_frame(
    name="toplam_tuketim_mwh"
)

yillik_analiz["degisim_yuzde"] = (
    yillik_analiz["toplam_tuketim_mwh"].pct_change() * 100
)

display(yillik_analiz.round(2))
```

```python
# Yıllık tüketim artışlarının incelenmesi
degisimler = yillik_analiz["degisim_yuzde"].dropna().to_numpy()

ortalama_artis = np.mean(degisimler)
standart_sapma = np.std(degisimler)

print(f"Ortalama yıllık artış: %{ortalama_artis:.2f}")
print(f"Yıllık artışların standart sapması: {standart_sapma:.2f} yüzde puan")
```

```python
# Yıllık elektrik tüketimi grafiği
plt.figure(figsize=(10, 5))

plt.plot(
    yillik_tuketim.index,
    yillik_tuketim.values / 1_000_000,
    marker="o",
    linewidth=2
)

plt.title("İstanbul'un Yıllık Elektrik Tüketimi (2020–2025)")
plt.xlabel("Yıl")
plt.ylabel("Elektrik Tüketimi (Milyon MWh)")
plt.xticks(yillik_tuketim.index)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Sektörlerin yıllara göre elektrik tüketimi
sektorel_yillik = (
    df_temiz[df_temiz["yil"] <= 2025]
    .pivot_table(
        index="yil",
        columns="tuketici_turu",
        values="elektrik_tuketimi_mwh",
        aggfunc="sum"
    )
)

display(sektorel_yillik.round(2))
```

```python
# Sektörlerin yıllık toplam tüketimdeki katkısı
ax = (sektorel_yillik / 1_000_000).plot(
    kind="bar",
    stacked=True,
    figsize=(11, 6),
    width=0.7
)

ax.set_title("İstanbul'da Sektörlere Göre Yıllık Elektrik Tüketimi")
ax.set_xlabel("Yıl")
ax.set_ylabel("Elektrik Tüketimi (Milyon MWh)")
ax.legend(title="Tüketici Grubu", bbox_to_anchor=(1.02, 1), loc="upper left")
ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Sektörlerin yıllık toplam tüketimdeki yüzde payları
sektorel_paylar = (
    sektorel_yillik.div(sektorel_yillik.sum(axis=1), axis=0) * 100
)

display(sektorel_paylar.round(2))
```

```python
# İstanbul'un aylık toplam elektrik tüketimi
aylik_tuketim = (
    df_temiz.groupby(["yil", "ay"])["elektrik_tuketimi_mwh"]
    .sum()
    .reset_index()
)

aylik_tuketim["tarih"] = pd.to_datetime(
    dict(
        year=aylik_tuketim["yil"],
        month=aylik_tuketim["ay"],
        day=1
    )
)

display(aylik_tuketim.head(12))
```

```python
# En yüksek ve en düşük tüketim aylarını bulma
tam_yillar = aylik_tuketim[
    aylik_tuketim["yil"] <= 2025
].reset_index(drop=True)

degerler = tam_yillar["elektrik_tuketimi_mwh"].to_numpy()

en_yuksek = tam_yillar.iloc[np.argmax(degerler)]
en_dusuk = tam_yillar.iloc[np.argmin(degerler)]

print(
    f"En yüksek tüketim: {int(en_yuksek['yil'])}-"
    f"{int(en_yuksek['ay']):02d} | "
    f"{en_yuksek['elektrik_tuketimi_mwh']:,.2f} MWh"
)

print(
    f"En düşük tüketim: {int(en_dusuk['yil'])}-"
    f"{int(en_dusuk['ay']):02d} | "
    f"{en_dusuk['elektrik_tuketimi_mwh']:,.2f} MWh"
)
```

```python
# Aylık elektrik tüketiminin zaman içindeki değişimi
plt.figure(figsize=(13, 5))

plt.plot(
    aylik_tuketim["tarih"],
    aylik_tuketim["elektrik_tuketimi_mwh"] / 1_000_000,
    marker="o",
    markersize=3
)

plt.title("İstanbul'un Aylık Elektrik Tüketimi (2020–Ağustos 2026)")
plt.xlabel("Tarih")
plt.ylabel("Elektrik Tüketimi (Milyon MWh)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Yıllara göre aylık toplam tüketim karşılaştırması
aylik_karsilastirma = aylik_tuketim.pivot(
    index="ay",
    columns="yil",
    values="elektrik_tuketimi_mwh"
)

display((aylik_karsilastirma / 1_000_000).round(2))
```

```python
# Yılların aylık tüketim eğrileri
plt.figure(figsize=(12, 6))

for yil in aylik_karsilastirma.columns:
    plt.plot(
        aylik_karsilastirma.index,
        aylik_karsilastirma[yil] / 1_000_000,
        marker="o",
        label=str(yil)
    )

plt.title("İstanbul'da Yıllara Göre Aylık Elektrik Tüketimi")
plt.xlabel("Ay")
plt.ylabel("Elektrik Tüketimi (Milyon MWh)")
plt.xticks(range(1, 13))
plt.legend(title="Yıl")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# 2020–2025 döneminde aylara göre ortalama elektrik tüketimi
aylik_ortalama = (
    aylik_tuketim[aylik_tuketim["yil"] <= 2025]
    .groupby("ay")["elektrik_tuketimi_mwh"]
    .agg(["mean", "median", "min", "max", "std"])
)

display(aylik_ortalama.round(2))
```

```python
# Aylara göre ortalama ve medyan tüketim karşılaştırması
plt.figure(figsize=(11, 5))

plt.plot(
    aylik_ortalama.index,
    aylik_ortalama["mean"] / 1_000_000,
    marker="o",
    label="Ortalama"
)

plt.plot(
    aylik_ortalama.index,
    aylik_ortalama["median"] / 1_000_000,
    marker="s",
    linestyle="--",
    label="Medyan"
)

plt.title("İstanbul'da Aylara Göre Elektrik Tüketimi (2020–2025)")
plt.xlabel("Ay")
plt.ylabel("Elektrik Tüketimi (Milyon MWh)")
plt.xticks(range(1, 13))
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Sektörlere göre aylık ortalama elektrik tüketimi (2020–2025)
sektorel_aylik_ortalama = (
    df_temiz[df_temiz["yil"] <= 2025]
    .groupby(["tuketici_turu", "ay"])["elektrik_tuketimi_mwh"]
    .mean()
    .unstack("tuketici_turu")
)

display(sektorel_aylik_ortalama.round(2))
```

```python
# Sektörlerin aylık ortalama tüketim eğrileri
fig, axes = plt.subplots(5, 1, figsize=(12, 15))

for ax, sektor in zip(axes, sektorel_aylik_ortalama.columns):
    ax.plot(
        sektorel_aylik_ortalama.index,
        sektorel_aylik_ortalama[sektor],
        marker="o"
    )

    ax.set_title(sektor)
    ax.set_ylabel("MWh")
    ax.set_xticks(range(1, 13))
    ax.grid(alpha=0.3)

axes[-1].set_xlabel("Ay")

plt.suptitle("Sektörlere Göre Aylık Ortalama Elektrik Tüketimi (2020–2025)")
plt.tight_layout()
plt.show()
```

```python
# 2020–2026 yıllarının Ocak–Ağustos tüketim toplamları
ilk_8_ay = (
    df_temiz[df_temiz["ay"] <= 8]
    .groupby("yil")["elektrik_tuketimi_mwh"]
    .sum()
    .to_frame(name="ocak_agustos_mwh")
)

# Önceki yılın aynı dönemine göre değişim
ilk_8_ay["degisim_yuzde"] = (
    ilk_8_ay["ocak_agustos_mwh"].pct_change() * 100
)

display(ilk_8_ay.round(2))
```

```python
# 2025 ve 2026'nın ilk sekiz ayını karşılaştırma
karsilastirma_2026 = (
    aylik_tuketim[
        (aylik_tuketim["yil"].isin([2025, 2026])) &
        (aylik_tuketim["ay"] <= 8)
    ]
    .pivot(index="ay", columns="yil", values="elektrik_tuketimi_mwh")
)

karsilastirma_2026["degisim_yuzde"] = (
    (karsilastirma_2026[2026] / karsilastirma_2026[2025] - 1) * 100
)

display(karsilastirma_2026.round(2))
```

```python
# Ocak–Ağustos toplam tüketim ve değişim
toplam_2025 = karsilastirma_2026[2025].sum()
toplam_2026 = karsilastirma_2026[2026].sum()

degisim = (toplam_2026 / toplam_2025 - 1) * 100

print(f"2025 Ocak–Ağustos: {toplam_2025:,.2f} MWh")
print(f"2026 Ocak–Ağustos: {toplam_2026:,.2f} MWh")
print(f"Değişim: %{degisim:.2f}")
```

```python
# 2025 ve 2026'nın ilk sekiz ayında sektörel tüketim
sektor_2026 = (
    df_temiz[
        (df_temiz["yil"].isin([2025, 2026])) &
        (df_temiz["ay"] <= 8)
    ]
    .pivot_table(
        index="tuketici_turu",
        columns="yil",
        values="elektrik_tuketimi_mwh",
        aggfunc="sum"
    )
)

sektor_2026["degisim_yuzde"] = (
    (sektor_2026[2026] / sektor_2026[2025] - 1) * 100
)

display(sektor_2026.round(2))
```

```python
# 2025–2026 Ocak–Ağustos sektörel değişim oranları
degisimler = sektor_2026["degisim_yuzde"].sort_values()

ax = degisimler.plot(
    kind="barh",
    figsize=(10, 5),
    color=["#d9534f" if x < 0 else "#2e8b57" for x in degisimler]
)

ax.axvline(0, color="black", linewidth=0.8)
ax.set_title("Sektörlere Göre Elektrik Tüketimindeki Değişim (2025–2026)")
ax.set_xlabel("Değişim (%)")
ax.set_ylabel("Tüketici Grubu")
ax.grid(axis="x", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Sektörlerin toplam tüketim değişimine katkısı
sektor_2026["fark_mwh"] = sektor_2026[2026] - sektor_2026[2025]

toplam_fark = sektor_2026["fark_mwh"].sum()

sektor_2026["katki_yuzde_puan"] = (
    sektor_2026["fark_mwh"] / sektor_2026[2025].sum() * 100
)

display(
    sektor_2026[
        ["fark_mwh", "degisim_yuzde", "katki_yuzde_puan"]
    ].round(2).sort_values("fark_mwh", ascending=False)
)

print(f"Toplam tüketim farkı: {toplam_fark:,.2f} MWh")
print(
    f"Toplam büyümeye katkıların toplamı: "
    f"{sektor_2026['katki_yuzde_puan'].sum():.2f} yüzde puan"
)
```

## 10. Elektrik Tüketimine Bağlı Karbon Emisyonlarının Hesaplanması

Bu bölümde İstanbul'un sektörel elektrik tüketimi, Enerji ve Tabii Kaynaklar Bakanlığı tarafından yayımlanan resmî emisyon faktörleri kullanılarak karbon emisyonu eşdeğerine dönüştürülecektir. Hesaplamalarda yıllara göre farklılaşan emisyon faktörleri dikkate alınacaktır.

Veri setinde tüketim noktalarının iletim veya dağıtım hattına bağlantı türü belirtilmediğinden, hesaplamalar iki ayrı senaryo üzerinden gerçekleştirilecektir. Elde edilen sonuçlar doğrudan ölçülmüş emisyonlar değil, elektrik tüketimine dayalı tahminlerdir.

**Veri kaynağı:** T.C. Enerji ve Tabii Kaynaklar Bakanlığı (ETKB), *Türkiye Elektrik Üretimi ve Elektrik Tüketim Noktası Emisyon Faktörleri* (2020–2023).

**Kaynak bağlantısı:** https://enerji.gov.tr/evced-cevre-ve-iklim-elektrik-uretim-tuketim-emisyon-faktorleri

**Veri kapsamı:** Türkiye geneli, yıllık emisyon faktörleri (tCO₂e/MWh). Hesaplamalarda iletim ve dağıtım faktörleri ayrı değerlendirilecektir.

```python
# ETKB resmî elektrik tüketim noktası emisyon faktörleri
# Birim: tCO2e/MWh

emisyon_faktorleri = pd.DataFrame({
    "yil": [2020, 2021, 2022, 2023],
    "iletim_faktoru": [0.427, 0.445, 0.445, 0.436],
    "dagitim_faktoru": [0.462, 0.479, 0.478, 0.469]
})

display(emisyon_faktorleri)
```

```python
# Eksik yıllar için 2023 faktörlerini referans alma
tum_faktorler = pd.DataFrame({"yil": range(2020, 2027)})

tum_faktorler = tum_faktorler.merge(
    emisyon_faktorleri,
    on="yil",
    how="left"
)

tum_faktorler["hesaplama_turu"] = tum_faktorler["yil"].apply(
    lambda yil: "Resmi faktor" if yil <= 2023
    else "2023 referans senaryosu"
)

tum_faktorler["iletim_faktoru"] = (
    tum_faktorler["iletim_faktoru"].fillna(
        emisyon_faktorleri.loc[
            emisyon_faktorleri["yil"] == 2023,
            "iletim_faktoru"
        ].iloc[0]
    )
)

tum_faktorler["dagitim_faktoru"] = (
    tum_faktorler["dagitim_faktoru"].fillna(
        emisyon_faktorleri.loc[
            emisyon_faktorleri["yil"] == 2023,
            "dagitim_faktoru"
        ].iloc[0]
    )
)

display(tum_faktorler)
```

```python
# Elektrik tüketimi ve emisyon faktörlerini birleştirme
df_karbon = df_temiz.merge(
    tum_faktorler,
    on="yil",
    how="left",
    validate="many_to_one"
)

# Birleştirme işlemini kontrol etme
print("Toplam gözlem sayısı:", len(df_karbon))

print("\nEksik emisyon faktörleri:")
print(
    df_karbon[
        ["iletim_faktoru", "dagitim_faktoru"]
    ].isna().sum()
)

print("\nHesaplama türlerine göre gözlem sayısı:")
print(df_karbon["hesaplama_turu"].value_counts())

display(df_karbon.head())
```

```python
# Elektrik tüketimine bağlı emisyon tahminlerini hesaplama
# MWh × tCO2e/MWh = tCO2e

df_karbon["emisyon_iletim_tco2e"] = (
    df_karbon["elektrik_tuketimi_mwh"]
    * df_karbon["iletim_faktoru"]
)

df_karbon["emisyon_dagitim_tco2e"] = (
    df_karbon["elektrik_tuketimi_mwh"]
    * df_karbon["dagitim_faktoru"]
)

# Hesaplanan değerleri inceleme
display(
    df_karbon[
        [
            "yil",
            "ay",
            "tuketici_turu",
            "elektrik_tuketimi_mwh",
            "hesaplama_turu",
            "emisyon_iletim_tco2e",
            "emisyon_dagitim_tco2e"
        ]
    ].head(10).round(2)
)
```

```python
# Yıllara göre toplam karbon emisyonları
yillik_karbon = (
    df_karbon.groupby("yil")[
        ["emisyon_iletim_tco2e", "emisyon_dagitim_tco2e"]
    ]
    .sum()
)

# Ton biriminden milyon ton birimine dönüştürme
yillik_karbon_mton = yillik_karbon / 1_000_000

display(yillik_karbon_mton.round(3))
```

```python
# Tamamlanmış yılların karbon emisyonları
karbon_grafik = yillik_karbon_mton.loc[2020:2025]

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(
    karbon_grafik.index,
    karbon_grafik["emisyon_iletim_tco2e"],
    marker="o",
    linewidth=2,
    label="İletim senaryosu"
)

ax.plot(
    karbon_grafik.index,
    karbon_grafik["emisyon_dagitim_tco2e"],
    marker="s",
    linewidth=2,
    label="Dağıtım senaryosu"
)

# Resmî faktörlerle referans senaryosunu ayırma
ax.axvline(2023.5, color="gray", linestyle="--", alpha=0.8)

ax.text(
    2021.5, 21.3, "Yıllık resmî faktörler",
    ha="center", fontsize=9
)

ax.text(
    2024.5, 21.3, "2023 referans faktörü",
    ha="center", fontsize=9
)

ax.set_title("İstanbul'un Elektrik Tüketimine Bağlı Tahmini Karbon Emisyonları")
ax.set_xlabel("Yıl")
ax.set_ylabel("Emisyon (milyon tCO₂e)")
ax.set_xticks(karbon_grafik.index)
ax.set_ylim(15, 22)
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Yıllara ve sektörlere göre karbon emisyonları
sektorel_karbon = (
    df_karbon[df_karbon["yil"] <= 2025]
    .pivot_table(
        index="yil",
        columns="tuketici_turu",
        values="emisyon_dagitim_tco2e",
        aggfunc="sum"
    )
)

# Milyon ton CO2 eşdeğerine dönüştürme
sektorel_karbon_mton = sektorel_karbon / 1_000_000

display(sektorel_karbon_mton.round(3))
```

```python
# Sektörlerin toplam karbon emisyonundaki yıllık payları (%)
karbon_paylari = (
    sektorel_karbon_mton.div(
        sektorel_karbon_mton.sum(axis=1),
        axis=0
    ) * 100
)

display(karbon_paylari.round(2))
```

```python
# Sektörlere göre yıllık karbon emisyonlarının değişimi
fig, ax = plt.subplots(figsize=(11, 6))

for sektor in sektorel_karbon_mton.columns:
    ax.plot(
        sektorel_karbon_mton.index,
        sektorel_karbon_mton[sektor],
        marker="o",
        linewidth=2,
        label=sektor
    )

# Resmî faktör dönemi ile referans senaryosunu ayırma
ax.axvline(
    2023.5,
    color="gray",
    linestyle="--",
    alpha=0.8
)

ax.set_title("Sektörlere Göre Tahmini Karbon Emisyonları (2020–2025)")
ax.set_xlabel("Yıl")
ax.set_ylabel("Emisyon (milyon tCO₂e)")
ax.set_xticks(sektorel_karbon_mton.index)
ax.legend(title="Tüketici Grubu")
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Yıllık elektrik tüketimi ve karbon emisyonu değişimi
tuketim_emisyon_karsilastirma = pd.DataFrame({
    "tuketim_degisim_yuzde":
        yillik_tuketim.pct_change() * 100,
    "emisyon_degisim_yuzde":
        yillik_karbon["emisyon_dagitim_tco2e"].pct_change() * 100
})

# Yalnızca yıllık resmî faktörlerin kullanıldığı dönem
display(
    tuketim_emisyon_karsilastirma.loc[2021:2023].round(2)
)
```

```python
# Tüketim ve emisyon değişimleri arasındaki fark
karsilastirma = tuketim_emisyon_karsilastirma.loc[
    2021:2023
].copy()

karsilastirma["fark_yuzde_puan"] = np.subtract(
    karsilastirma["emisyon_degisim_yuzde"],
    karsilastirma["tuketim_degisim_yuzde"]
)

display(karsilastirma.round(2))
```

```python
# 2025 ve 2026 Ocak–Ağustos karbon emisyonları
karbon_2026 = (
    df_karbon[
        (df_karbon["yil"].isin([2025, 2026])) &
        (df_karbon["ay"] <= 8)
    ]
    .groupby("yil")[
        ["emisyon_iletim_tco2e", "emisyon_dagitim_tco2e"]
    ]
    .sum()
)

# Milyon ton CO2 eşdeğerine dönüştürme
karbon_2026_mton = karbon_2026 / 1_000_000

# İki dönem arasındaki değişim
degisim_karbon = (
    (karbon_2026.loc[2026] / karbon_2026.loc[2025] - 1) * 100
)

display(karbon_2026_mton.round(3))

print("\n2025–2026 Ocak–Ağustos emisyon değişimi (%):")
display(degisim_karbon.round(2))
```

**3. Sektörel Karbon Emisyonları**

- Bütün sektörlere aynı yıl için ortak emisyon faktörü uygulandığından, sektörel emisyon dağılımı elektrik tüketiminin dağılımıyla aynı oransal yapıya sahiptir.
- Sektörlerin toplam tahmini emisyon içindeki payları elektrik tüketim paylarıyla aynı dağılımı göstermektedir. 2025 yılında en yüksek pay %41,37 ile Hizmetler/Ticarethane grubuna, ardından %31,53 ile mesken ve %25,93 ile sanayi gruplarına aittir. Bu sonuç, tüm sektörler için aynı yıllık emisyon faktörünün kullanılmasının matematiksel bir sonucudur.

**4. 2025–2026 Ocak–Ağustos Karşılaştırması**
- Dağıtım senaryosunda tahmini emisyon 2025'in ilk sekiz ayında 14,278 milyon tCO₂e, 2026'nın aynı döneminde 14,619 milyon tCO₂e olarak hesaplanmıştır.
- Her iki bağlantı senaryosunda da emisyon artış oranı %2,39'dur.
- Her iki dönemde aynı 2023 referans faktörü kullanıldığından bu artış, doğrudan elektrik tüketimindeki %2,39'luk artışı yansıtmaktadır.

**5. Hesaplamanın Sınırlılıkları**
- Hesaplanan değerler elektrik tüketimine bağlı dolaylı sera gazı emisyonu tahminleridir; sektörlerin doğrudan yakıt tüketimi ve üretim süreçlerinden kaynaklanan emisyonları kapsanmamaktadır.
- Elektrik tüketim verilerinde bağlantı türü ayrımı bulunmadığından iletim ve dağıtım faktörleri iki ayrı senaryo olarak değerlendirilmiştir.
- 2024–2026 dönemindeki emisyonlar, 2023 yılı faktörlerinin sabit kaldığı varsayımına dayanmaktadır.
- Yıllar arasında değişen sektörel tüketici sınıflandırmaları, uzun dönemli sektörel karşılaştırmaların yorumlanmasında dikkate alınmalıdır.

**Genel Değerlendirme**

Analizler, elektrik tüketimindeki artışın her zaman aynı oranda karbon emisyonu artışı anlamına gelmediğini göstermektedir. Hesaplanan emisyon miktarı hem tüketim düzeyine hem de kullanılan emisyon faktörüne bağlıdır. Özellikle 2023 yılındaki sonuçlar bu ilişkinin önemini ortaya koymaktadır. Bununla birlikte 2024–2026 dönemi için elde edilen değerler, resmî yıllık emisyon envanteri değil, referans senaryosu sonuçları olarak değerlendirilmelidir.

## 11. Genel Sonuç ve Öneriler

Bu çalışmada İstanbul'un 2020–Ağustos 2026 dönemindeki sektörel elektrik tüketimi, EPDK verileri kullanılarak analiz edilmiş; elektrik tüketimine bağlı dolaylı sera gazı emisyonları ise resmî emisyon faktörleri ve referans senaryoları aracılığıyla tahmin edilmiştir.

### Genel Sonuçlar

- **Tüketim eğilimi:** İstanbul'un elektrik tüketimi 2020–2025 döneminde yıllık bazda sürekli artış göstermiştir. Ancak artış oranlarının yıllar arasında farklılaştığı belirlenmiştir.
- **Sektörel dağılım:** Hizmetler/Ticarethane grubu, toplam elektrik tüketimi ve hesaplanan elektrik kaynaklı emisyonlar içerisinde en yüksek paya sahiptir. Bu grubu mesken ve sanayi takip etmektedir.
- **Dönemsel farklılıklar:** Aylık ve mevsimsel analizlerde tüketimin yıl boyunca sabit olmadığı, sektörlere göre farklı tüketim eğilimlerinin bulunduğu görülmüştür.
- **Veri kalitesi:** Eksik veya tekrarlanan kayıt tespit edilmemiş; aykırı gözlemler incelenerek veri seti içerisinde korunmuştur. Yıllar arasında değişen tüketici grubu adları analiz amacıyla standartlaştırılmıştır.
- **Tüketim ve emisyon ilişkisi:** Elektrik tüketimindeki artışın her zaman hesaplanan emisyonlarda aynı yönde bir değişim oluşturmadığı görülmüştür. Yıllık emisyon faktörlerindeki farklılıklar emisyon tahminlerini etkilemektedir.
- **2026 değerlendirmesi:** Ocak–Ağustos dönemi karşılaştırmasında 2026 tüketimi, 2025'in aynı döneminden yüksek gerçekleşmiştir. Sabit referans faktörlü emisyon senaryoları da aynı yönde değişmiştir.

### Öneriler ve Çalışmanın Sınırlılıkları

- Tüketim değişimlerinin olası nedenlerini incelemek için sıcaklık, nüfus ve ekonomik faaliyet verileriyle analizler genişletilebilir.
- Sektörlerin gerçek karbon yoğunluklarının karşılaştırılması, elektrik tedarik özelliklerini ve sektöre özgü emisyon verilerini gerektirmektedir.
- 2024–2026 dönemindeki emisyon sonuçları, 2023 faktörlerine dayalı varsayımsal hesaplamalardır; bu dönemlerin doğrulanmış yıllık emisyon değerleri olarak değerlendirilmemelidir.
- Tüketici gruplarının yıllar arasındaki kapsam farklılıkları, uzun dönemli sektörel sonuçların yorumlanmasında dikkate alınmalıdır.

**Genel olarak**, çalışma İstanbul'un sektörel elektrik tüketiminin zaman içindeki değişimini ortaya koymakta ve tüketim verilerinin dolaylı sera gazı emisyonu tahminlerinde nasıl kullanılabileceğini göstermektedir.
